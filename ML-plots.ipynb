```python
import matplotlib.pyplot as plt
```

# Variance VS Components Kept

```python
var=[0.1, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4, 0.5,0.55, 
     0.6,0.65,0.7,0.75,0.8,0.85,0.9,0.95,0.96,]

Components=[1,1,1,1,2,2,2,4,5,7,10,15,22,33,52,87,178,272]
```

```python
fig, ax = plt.subplots()  # Create a figure containing a single axes.
ax.plot(var, Components);  # Plot some data on the axes.
ax.set_xlabel('Variance')
ax.set_ylabel('# of Components kept')
ax.set_title('Variance VS Components')
plt.show()
```
